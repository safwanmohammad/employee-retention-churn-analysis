## Missing Values Handling 

Identify missing values and understand whether they are:
- Data errors
- Business-related missing values

```python
for col in df:
    print(f"---------{col}----------")
    print(df[col].unique())
```

```python
# Replace job_id with numbers starting from 1 to N
df['job_id'] = np.arange(1 , len(df) + 1)
```

```python
#  clean Jan-50 to 1-50 in company_size
# df['company_size'] = df['company_size'].replace('Jan-50',"1-50",regex=True)
df['company_size'] = df['company_size'].replace('(?i)Jan-50', "1-50", regex=True)
```

```python
# fixing ~ education_required not null to 'Not-specified'
df['education_required'] = df['education_required'].fillna('Not-specified')
```

```python
# converting string datatype (application_deadline and posted_date) to date format.
df['application_deadline'] = pd.to_datetime(df['application_deadline'],dayfirst=True,errors='coerce')
```

```python
df['posted_date'] = pd.to_datetime(df['posted_date'],dayfirst=True,errors='coerce')
```

```python
# in avg_response_time convarting negative value to absolute value 
# also change the feature name into avg_response_hours
df = df.rename(columns={'avg_response_time':'avg_response_hours'})
df['avg_response_hours'] = df['avg_response_hours'].abs()
```

```python
# grouping avg_response_time in a 4 group

df['avg_response_hours'] = pd.cut(
    df['avg_response_hours'], 
    bins=[-float('inf'), 6, 12, 18, float('inf')], 
    labels=['0-6', '6-12', '12-18', '18-24']
)
```

```python
# starting_salary 
df['starting_salary'].isna().mean()* 100   #about 50% datas are missing
```

```python
# create a new feature company type 
company_type_mapping = {
    "1-50": "Startup",
    "51-200": "Scale-up",
    "201-500": "Mid-Market",
    "501-1000": "Enterprise",
    "1001-5000": "Corporation",
    "5001-10000": "MNC",
    "10000+": "MNC",
}
df["company_tier"] = df["company_size"].map(
    company_type_mapping
)
```
